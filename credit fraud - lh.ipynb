```python
from sklearn import linear_model
from sklearn import tree
from sklearn import neighbors
```

```python
clf = linear_model.LogisticRegression()
```

```python
def model_validation(model, X_train_scaled, X_test_scaled, y_train, y_test):

  clf = model
  clf.fit(X_train_scaled, y_train)

  df_validation = pd.DataFrame()
  df_validation["Amount"] = X_test['Amount']
  df_validation["Class"] = y_test
  df_validation["predicted"] = clf.predict(X_test_scaled)

  df_loses = df_validation[df_validation['Class'] == 1]
  total_loss = df_loses['Amount'].sum()

  mask = (df_loses['predicted'] == 0) & (df_loses['Class'] == 1)
  model_loss = df_loses[mask]['Amount'].sum()

  return model_loss/total_loss, clf.score(X_test_scaled, y_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
classifiers = {
    "LogisiticRegression": LogisticRegression(random_state=42),
    "KNearest": KNeighborsClassifier(),
    "Support Vector Classifier": SVC(random_state=42),
    "DecisionTreeClassifier": DecisionTreeClassifier(random_state=42),
    "RandomForestClassifier": RandomForestClassifier(class_weight='balanced', random_state=42)
}
```

```python
X, y = df.drop(columns =['Class'], axis=1), df['Class']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
scaler = StandardScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
for key, classifier in classifiers.items():
  print(key + ":")
  print(model_validation(classifier, X_train_scaled, X_test_scaled, y_train, y_test))
```

## Zadanie: stworzyć najlepszy model korzystająć z grid-search i utworzyć projekt w pycharm (load_data, train, predict)

```python
# najlepszy model korzystająć z grid-search

from sklearn.model_selection import GridSearchCV

# Define the parameter grid for RandomForestClassifier
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# Create a RandomForestClassifier instance
rf_classifier = RandomForestClassifier(random_state=42, class_weight='balanced')

# Perform GridSearchCV
grid_search = GridSearchCV(estimator=rf_classifier, param_grid=param_grid, cv=5, scoring='f1', n_jobs=-1) # Use f1 scoring for imbalanced datasets
grid_search.fit(X_train_scaled, y_train)

# Get the best model and its score
best_rf_classifier = grid_search.best_estimator_
best_score = grid_search.best_score_

print(f"Best RandomForestClassifier parameters: {grid_search.best_params_}")
print(f"Best f1 score: {best_score}")


# Evaluate the best model
loss, accuracy = model_validation(best_rf_classifier, X_train_scaled, X_test_scaled, y_train, y_test)
print(f"Loss ratio for best model: {loss}")
print(f"Accuracy for best model: {accuracy}")
```

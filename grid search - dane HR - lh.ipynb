```python
import pandas as pd
```

```python
df = pd.read_csv('HR_capstone_dataset.csv')
```

```python
df.head()
```

```python
df_dummy = pd.get_dummies(df, drop_first=True)
df_dummy
```

```python
from sklearn.model_selection import train_test_split

X = df_dummy.drop(columns=['left'], axis=1)
y = df_dummy['left']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn import tree
dtc_tree = tree.DecisionTreeClassifier(random_state=0)
dtc_tree.fit(X_train, y_train)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, dtc_tree.predict(X_test)))
```

```python
print(classification_report(y_train, dtc_tree.predict(X_train)))
```

```python
dtc_tree = tree.DecisionTreeClassifier(random_state=0, min_samples_leaf=50, max_depth=3)
dtc_tree.fit(X_train, y_train)
print(classification_report(y_test, dtc_tree.predict(X_test)))
```

```python
from sklearn import svm, datasets
from sklearn.model_selection import GridSearchCV
iris = datasets.load_iris()
```

```python
iris.target
```

```python
iris.data
```

```python
parameters = {'kernel':('linear', 'rbf'), 'C':[1, 10]}
svc = svm.SVC(random_state=0)
clf = GridSearchCV(svc, parameters)
clf.fit(iris.data, iris.target)
sorted(clf.cv_results_.keys())
```

```python
clf.cv_results_
```

```python
clf.best_estimator_  # wynik szukania najlepszych parametów
```

# Dobór hiper parametów

```python
parameters = {
    'min_samples_leaf':[200, 500, 600],
    'max_depth':[5, 10, 15, 20, 25]
    }
```

## Zadanie dobrać parametru modelu korzystając z gridsearch

```python
dtc_tree = tree.DecisionTreeClassifier(random_state=0)
clf = GridSearchCV(dtc_tree, parameters)
clf.fit(X_train, y_train)
```

```python
clf.best_estimator_
```

```python
clf.best_params_
```

```python
clf.best_score_
```

```python
tree_sel = clf.best_estimator_
```

```python
tree_sel.fit(X_train, y_train)
```

```python
print(classification_report(y_test, tree_sel.predict(X_test)))
```

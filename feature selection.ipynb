```python
import pandas as pd
```

```python
df = pd.read_csv('creditcard.csv').sample(50000, random_state=1)
```

```python
from sklearn.feature_selection import SequentialFeatureSelector
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
```

```python
X, y = df.drop(columns =['Class'], axis=1), df['Class']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
scaler = StandardScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

sklearn.feature_selection.SequentialFeatureSelector  `(estimator, *, n_features_to_select='auto', tol=None, direction='forward', scoring=None, cv=5, n_jobs=None)`

```python
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)
knn = KNeighborsClassifier(n_neighbors=3)
sfs = SequentialFeatureSelector(knn, n_features_to_select=3)
sfs.fit(X, y)
sfs.get_support()
sfs.transform(X).shape
```

```python
clf_tree = DecisionTreeClassifier(random_state=42)
```

```python
sfs = SequentialFeatureSelector(clf_tree, n_features_to_select=3, direction='forward')
```

```python
sfs.fit(X, y)
```

```python
sfs.get_support()
```

```python
sfs.get_feature_names_out()
```
